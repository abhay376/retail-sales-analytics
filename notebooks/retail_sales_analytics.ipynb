## Step 1: Data Ingestion & Inspection (Pandas & NumPy)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3

# Load raw dataset
df = pd.read_csv('../data/retail_sales_raw.csv')
print(f"Dataset Shape: {df.shape}")
df.head()
```

## Step 2: Data Cleaning & Feature Transformations

```python
# Check for missing values and data types
print(df.isnull().sum())

# Feature engineering: Converting Date and extracting temporal features
df['Date'] = pd.to_datetime(df['Date'])
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['YearMonth'] = df['Date'].dt.to_period('M')
df['DayOfWeek'] = df['Date'].dt.day_name()
df['IsWeekend'] = df['Date'].dt.dayofweek.isin([5, 6]).astype(int)

df.describe()
```

## Step 3: Visual Exploratory Data Analysis (Matplotlib & Seaborn)

```python
# Plot 1: Revenue by Product Category
plt.figure(figsize=(10, 5))
cat_revenue = df.groupby('Product_Category')['Total_Amount'].sum().sort_values(ascending=False).reset_index()
sns.barplot(data=cat_revenue, x='Product_Category', y='Total_Amount', hue='Product_Category', palette='mako', legend=False)
plt.title('Total Revenue by Product Category', fontsize=14, fontweight='bold')
    plt.xlabel('Category')
plt.ylabel('Revenue ($)')
plt.show()
```

```python
# Plot 2: Monthly Sales Trend Line Chart
monthly = df.groupby('YearMonth')['Total_Amount'].sum().reset_index()
monthly['YearMonth'] = monthly['YearMonth'].dt.to_timestamp()

plt.figure(figsize=(12, 5))
plt.plot(monthly['YearMonth'], monthly['Total_Amount'], marker='o', color='#2b5c8f', linewidth=2)
plt.title('Monthly Sales Revenue Trend', fontsize=14, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## Step 4: SQL Analysis & Queries (PostgreSQL / SQLite Integration)

```python
# Connect to SQLite Database
conn = sqlite3.connect('../data/retail_database.db')

# SQL Query: Sales Performance & Unique Customers by Location
query = '''
SELECT 
    Store_Location,
    COUNT(Transaction_ID) AS Total_Transactions,
    SUM(Total_Amount) AS Total_Revenue,
    ROUND(AVG(Total_Amount), 2) AS Avg_Order_Value
FROM retail_sales
GROUP BY Store_Location
ORDER BY Total_Revenue DESC;
'''

sql_result = pd.read_sql_query(query, conn)
conn.close()
sql_result
```

## Step 5: Demand Forecasting Model (Scikit-Learn)

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score

# Aggregate daily sales
daily = df.groupby('Date')['Total_Amount'].sum().reset_index()
daily['DayOfYear'] = daily['Date'].dt.dayofyear
daily['DayOfWeek'] = daily['Date'].dt.dayofweek
daily['Month'] = daily['Date'].dt.month
daily['Year'] = daily['Date'].dt.year
daily['Sales_Lag_1'] = daily['Total_Amount'].shift(1)
daily['Rolling_Avg_7'] = daily['Total_Amount'].shift(1).rolling(7).mean()

model_df = daily.dropna()
X = model_df[['DayOfYear', 'DayOfWeek', 'Month', 'Year', 'Sales_Lag_1', 'Rolling_Avg_7']]
y = model_df['Total_Amount']

# Train / Test split (80-20)
split = int(len(model_df) * 0.8)
X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y.iloc[:split], y.iloc[split:]

model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

preds = model.predict(X_test)
print(f"MAE: ${mean_absolute_error(y_test, preds):.2f}")
print(f"R² Score: {r2_score(y_test, preds):.4f}")
```
